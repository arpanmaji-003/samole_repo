```python
#collection: list,trple, set, dict
#list : list, mutable, allow duplicates
fruits = ['apple','orange','mango','pineapple']
fruits[2] = "watermalon" # update the existing element
fruits.append('blueberry') # append always add element last of the list
fruits.insert(2, 'banana') # insert add element any place i want ion the list
fruits.extend(['kiwi', 'papaya','graphs']) # extend for add number of element in the list
fruits.remove('apple') # remove apple from the list
fruits.sort(reverse=True) # sort the list in accending order and return none
print(fruits)
print(type(fruits))
print(len(fruits))

print(fruits[1])
print(fruits[-1]) # if we do not know number of list then use -1 reverse for print the last element
print(fruits[1:3])
print(fruits[1:])
print(fruits[:3])
```

```python
num = float(input('Enter a Number : '))
if num > 0:
  print("Number is positive")
elif num < 0:
  print('Number is negative')
else:
  print('Number is neutral')
```

```python
# while loop
lucky_number=3000
while True:
  number=int(input('Enter a Number: '))
  if number>999 and number <=10000:
   if number==lucky_number:
    print('Congragulation! You have gauss the correct number')
    break
   else:
    print('Try again!')
  else:
    print('Enter 4 digits number.')
```

```python
# take two number from user and if the product is greater then 5000 print 'pass' else 'keep try'
while True:
  onumber=int(input('Enter 1st Number: '))
  snumber = int(input('Enter 2nd Number: '))
  if onumber>0 and snumber>0:
    if onumber*snumber>5000:
      print('PASS')
      break
    else:
      print('KEEP TRY')
  else:
    print('NUmber should be greater then 0')
```
